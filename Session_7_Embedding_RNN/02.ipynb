```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split

import tensorflow as tf

from tensorflow.keras import layers, models
```

```python
df = pd.read_csv("imdb_reviews.csv")
df
```

```python
df["target"] = df["sentiment"].map({"positive": 1, "negative": 0})

X = df["review"].to_numpy()
y = df["target"].to_numpy()
```

```python
X_len = [len(x) for x in X]
np.mean(X_len)
```

```python
np.quantile(X_len, 0.9)
```

```python
x_train, x_test, y_train, y_test = train_test_split(X,y, train_size=0.8, stratify=y)
```

```python
vectorizer = layers.TextVectorization(
    max_tokens=10000,
    output_sequence_length=200
)

vectorizer.adapt(x_train)
```

```python
model = models.Sequential([
    layers.Input(shape=(), dtype=tf.string),

    vectorizer,

    layers.Embedding(
        input_dim=10000,
        output_dim=200,
        mask_zero=True
    ),

    layers.Bidirectional(layers.SimpleRNN(32, return_sequences=True)),
    layers.SimpleRNN(32),


    layers.Dense(1, activation="sigmoid")

])

model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
model.fit(
    x_train,
    y_train,
    epochs=5,
    batch_size=32,
    validation_split=0.2,
)
```

```python
model.evaluate(x_test, y_test)
```

```python
# Vocabulary
vocab = vectorizer.get_vocabulary()

# Embedding weights: (10000, 16)
embedding_weights = model.layers[1].get_weights()[0]

# Dense weights: (16, 1)
dense_weights = model.layers[-1].get_weights()[0]

# Calculate word scores
scores = embedding_weights @ dense_weights[:, 0]

# Create DataFrame (exclude Padding and OOV)
df_words = pd.DataFrame({
    "word": vocab[2:],
    "score": scores[2:len(vocab)]
})

# Top 10 positive words
positive_words = df_words.nlargest(30, "score")

# Top 10 negative words
negative_words = df_words.nsmallest(30, "score")

print("Positive Words:")
print(positive_words)

print("\nNegative Words:")
print(negative_words)
```
