```python
import numpy as np
from keras.src.layers import Conv2DTranspose
from matplotlib import pyplot as plt
import seaborn as sns

from sklearn.metrics import confusion_matrix

from tensorflow.keras.models import Sequential

from tensorflow.keras.layers import (
    Input,
    Conv2D,
    MaxPooling2D,
    BatchNormalization,
    Dropout,
    Dense,
    GlobalAveragePooling2D,
    SeparableConv2D,
    Conv2DTranspose,
    DepthwiseConv2D,
    UpSampling2D
)

from tensorflow.keras.datasets import cifar10
```

```python
(x_train, y_train), (x_test, y_test) = cifar10.load_data()

x_train.shape, x_test.shape
```

```python
model = Sequential([

    Input(shape=(32,32,3)),

    Conv2DTranspose()
    SeparableConv2D(32,(3,3),padding="same",activation="relu"),
    BatchNormalization(),
    SeparableConv2D(32,(3,3),padding="same",activation="relu"),
    BatchNormalization(),
    MaxPooling2D((2,2)),
    Dropout(0.25),

    SeparableConv2D(64,(3,3),padding="same",activation="relu"),
    BatchNormalization(),
    SeparableConv2D(64,(3,3),padding="same",activation="relu"),
    BatchNormalization(),
    MaxPooling2D((2,2)),
    Dropout(0.3),

    SeparableConv2D(128,(5,5),padding="same",activation="relu"),
    BatchNormalization(),
    SeparableConv2D(128,(3,3),padding="same",activation="relu"),
    BatchNormalization(),
    MaxPooling2D((2,2)),
    Dropout(0.4),

    GlobalAveragePooling2D(),

    Dense(10,activation="softmax")

])

model.summary()
```

```python
model.compile(
    loss="sparse_categorical_crossentropy",
    optimizer="adam",
    metrics=["accuracy"],
)
```

```python
history = model.fit(
    x_train,
    y_train,
    validation_split=0.2,
    batch_size=64,
    epochs=2
)

history = history.history
```

```python
history["accuracy"]
history["val_accuracy"]

history["loss"]
history["val_loss"]
```

```python
# 28,28,1 => 3,3,32 ==> 28,28,32 ==> 14,14,32
# 14,14,32 => 3,3,32 ==> 14,14,32 ==> 7,7,32
```

```python
model.evaluate(x_test, y_test)
```
