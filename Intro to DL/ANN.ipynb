مفهوم دوم: Graph (گراف محاسباتی) – نقشه راه اجرای عملیات‌ها
در نسخه‌های اولیه TensorFlow (و همچنان در پشت پرده TensorFlow 2.x)، هسته مرکزی طراحی، مفهوم “گراف محاسباتی” (Computational Graph) بود. یک گراف محاسباتی، یک نمایش انتزاعی از تمام عملیات‌هایی است که قرار است روی داده‌ها انجام شود. این گراف از گره‌ها (Nodes) و یال‌ها (Edges) تشکیل شده است: گره‌ها نشان‌دهنده عملیات‌های ریاضی (Operations یا Ops) هستند و یال‌ها جریان داده‌ها را در قالب تانسورها (Tensors) بین عملیات‌ها نشان می‌دهند.

مزایای اصلی استفاده از گراف‌ها در TensorFlow عبارتند از:

بهینه‌سازی عملکرد: TensorFlow می‌تواند گراف را قبل از اجرا تحلیل و بهینه‌سازی کند. این بهینه‌سازی‌ها شامل حذف عملیات‌های غیرضروری، ادغام عملیات‌ها و برنامه‌ریزی کارآمد اجرای آن‌ها می‌شود.
توزیع‌پذیری: گراف‌ها را می‌توان به راحتی بر روی چندین CPU، GPU یا حتی چندین سرور در یک محیط توزیع‌شده اجرا کرد، که برای مدل‌های بزرگ و حجیم بسیار مهم است.
صادرات و استقرار: یک گراف آموزش‌دیده را می‌توان ذخیره کرد (سریالایز کرد) و سپس در محیط‌های مختلف مانند TensorFlow Lite برای دستگاه‌های موبایل یا TensorFlow.js برای وب، بدون نیاز به کد پایتون اصلی بارگذاری و اجرا نمود.
محاسبه خودکار گرادیان‌ها: TensorFlow از ساختار گراف برای انجام مشتق‌گیری خودکار استفاده می‌کند که برای الگوریتم‌های بهینه‌سازی مانند گرادیان کاهشی (Gradient Descent) در آموزش مدل‌ها ضروری است.
در مدل گراف-محور TensorFlow 1.x، شما ابتدا باید تمام محاسبات خود را به صورت یک گراف تعریف می‌کردید و سپس برای اجرای آن و به دست آوردن مقادیر واقعی، از یک Session استفاده می‌کردید. این جدایی بین فاز “تعریف” و فاز “اجرا”، اگرچه قدرتمند بود، اما برای توسعه‌دهندگان جدید پیچیدگی‌هایی ایجاد می‌کرد.

با ورود TensorFlow 2.x و معرفی Eager Execution به عنوان حالت پیش‌فرض، نیازی به ساخت صریح گراف و اجرای آن با Session نیست. در این حالت، عملیات‌ها به محض فراخوانی اجرا می‌شوند، درست مانند پایتون معمولی. با این حال، مفهوم گراف هنوز در پس‌زمینه برای بهینه‌سازی عملکرد (هنگام استفاده از `tf.function` برای کامپایل کد به گراف) و استقرار مدل‌ها وجود دارد.



مفهوم سوم: Session – موتور اجرایی گراف در گذشته TensorFlow
در TensorFlow 1.x و در چارچوب برنامه‌نویسی گراف‌محور، “Session” (جلسه) نقش حیاتی یک محیط زمان اجرا (runtime environment) را ایفا می‌کرد. یک Session مسئول تخصیص منابع سخت‌افزاری (مانند CPU و GPU)، مدیریت وضعیت متغیرهای گراف و اجرای عملیات تعریف شده در گراف محاسباتی بود. بدون یک Session، گرافی که شما تعریف می‌کردید، تنها یک طرح کلی باقی می‌ماند و هیچ محاسبه‌ای واقعاً انجام نمی‌شد.

برای فعال کردن محاسبات و دسترسی به نتایج تانسورها، باید گرافی که ساخته‌اید را به متد `run()` یک Session ارسال می‌کردید. این رویکرد، کنترل دقیقی بر نحوه و زمان اجرای محاسبات فراهم می‌کرد، اما در عین حال، لایه‌ای از پیچیدگی را به فرآیند توسعه اضافه می‌نمود.

نحوه استفاده از Session در TensorFlow 1.x به شکل زیر بود (توجه داشته باشید که این کد برای TF 1.x است):


همانطور که می‌بینید، متغیر `c` تنها یک تانسور است که عملیات آن در گراف تعریف شده و مقدار نهایی آن فقط پس از فراخوانی `sess.run(c)` محاسبه می‌شود. این الگوی “ساخت و سپس اجرا” ویژگی اصلی TensorFlow 1.x بود.

با معرفی TensorFlow 2.x و پذیرش Eager Execution به عنوان پیش‌فرض، نیاز به ایجاد و مدیریت صریح Session برای اکثر کاربران به طور قابل توجهی کاهش یافته است. عملیات‌ها بلافاصله پس از فراخوانی اجرا می‌شوند و نتایج فوراً در دسترس هستند، که این امر توسعه، اشکال‌زدایی و درک کد را به میزان زیادی ساده کرده است. با این حال، همانطور که قبلاً اشاره شد، مکانیزم‌های اجرایی مبتنی بر گراف هنوز در پس‌زمینه برای بهینه‌سازی و استقرار مدل‌ها، به ویژه هنگام استفاده از `tf.function`، فعال هستند.

````python
import tensorflow.compat.v1 as tf
# 1. Define the computational graph
a = tf.constant(7.0, name="input_a")
b = tf.constant(8.0, name="input_b")
c = tf.add(a, b, name="add_operation")

# 2. Create a Session and run the graph
with tf.Session() as sess:

    # Execute the graph to get the value of 'c'
    result = sess.run(c)
    print(f"Result of c: {result}") # Output: Result of c: 15.0
# In TF 1.x, without sess.run(), 'c' is just a Tensor object, not a value.
# print(c) # Would output: Tensor("add_operation:0", shape=(), dtype=float32)


````

 نسخه‌های قدیمی‌تر TensorFlow، این سه مفهوم – Tensor، Graph و Session – به طور جدایی‌ناپذیری به هم گره خورده بودند. شما با تانسورها داده‌ها را ساختار می‌دادید، با استفاده از عملیات‌هایی که تانسورها را به هم متصل می‌کردند، یک گراف محاسباتی ایجاد می‌کردید، و سپس برای اینکه این گراف زنده شود و محاسبات انجام شود، از یک Session استفاده می‌نمودید. این رویکرد یکپارچه، اگرچه برای مقیاس‌پذیری و بهینه‌سازی‌های پیچیده بسیار کارآمد بود، اما برای بسیاری از توسعه‌دهندگان جدید و حتی باتجربه، نقطه ورود دشواری محسوب می‌شد و فرآیند اشکال‌زدایی را طولانی‌تر می‌کرد.

TensorFlow 2.x با هدف ساده‌سازی تجربه کاربری و هماهنگی بیشتر با فلسفه برنامه‌نویسی پایتون، تحول عظیمی را با معرفی Eager Execution به عنوان حالت پیش‌فرض به ارمغان آورد. در این پارادایم جدید:

تانسورها: همچنان ستون فقرات داده‌ها هستند، اما مقادیر آن‌ها بلافاصله پس از تعریف و انجام عملیات، مانند آرایه‌های NumPy، در دسترس قرار می‌گیرند.
گراف‌ها: دیگر نیازی به ساخت صریح گراف نیست. TensorFlow به طور هوشمند و در پس‌زمینه، کدهای پایتون را به گراف‌های قابل اجرا تبدیل می‌کند، به ویژه زمانی که از `tf.function` برای بهینه‌سازی عملکرد استفاده می‌شود.
Session: از API عمومی حذف شده است. TensorFlow به صورت خودکار محیط زمان اجرا را مدیریت می‌کند و کاربران نیازی به تعامل مستقیم با Session ندارند.
این تغییرات، TensorFlow را به ابزاری منعطف‌تر، کاربرپسندتر و قابل فهم‌تر برای برنامه‌نویسان پایتون تبدیل کرده است، در حالی که همچنان از قدرت و کارایی بی‌نظیر گراف‌های محاسباتی برای مقیاس‌پذیری و عملکرد بالا بهره می‌برد. این رویکرد جدید، فاصله بین کدنویسی تعاملی و تولیدی را کمتر می‌کند.

## 6. Computational graphs
A computational graph represents operations

https://www.tensorflow.org/guide/intro_to_graphs

Graphs are data structures that contain a set of tf.Operation objects, which represent units of computation; and tf.Tensor objects, which represent the units of data that flow between operations. They are defined in a tf.Graph context. Since these graphs are data structures, they can be saved, run, and restored all without the original Python code.

With a graph, you have a great deal of flexibility. You can use your TensorFlow graph in environments that don't have a Python interpreter, like mobile applications, embedded devices, and backend servers. TensorFlow uses graphs as the format for saved models when it exports them from Python.

Graphs are also easily optimized, allowing the compiler to do transformations like:

Statically infer the value of tensors by folding constant nodes in your computation ("constant folding").
Separate sub-parts of a computation that are independent and split them between threads or devices.
Simplify arithmetic operations by eliminating common subexpressions.
There is an entire optimization system, Grappler, to perform this and other speedups.

In short, graphs are extremely useful and let your TensorFlow run fast, run in parallel, and run efficiently on multiple devices.

However, you still want to define your machine learning models (or other computations) in Python for convenience, and then automatically construct graphs when you need them.

````python
import numpy as np

X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([
    0,
    1,
    1,
    1
])
````

````python
# 2. Initialize weights and biases
np.random.seed(42)

# for hidden layer we have 2 neurons and each gets 2 input (as we have 2 feature per sample) so need (2,2) Weights means each neuron need 2 w
W1 = np.random.randn(2,2)
b1 = np.zeros((1,2))

W2 = np.random.randn(2,1)
b2 = np.zeros((1,1))
W1
````

````python
# 3. Forward propagation
Z1 = X @ W1 + b1
Z1
# (4,2) @ (2,2)
#       ↓
#     (4,2)  4 samples each get 2 values as we have 2 neurons
# this is like mini batch
# columns 1 is in fact the first hidden neuron and columns 2 is the second hidden neuron calculations
````

````python
# 4. Apply activation
def relu(z):
    return np.maximum(0,z)
A1 = relu(Z1)
A1
````

````python
# 5. Calculate output layer
Z2 = A1 @ W2 + b2
Z2
````

````python
# 6. Sigmoid
def sigmoid(z):
    return 1 / (1+np.exp(-z))

A2 = sigmoid(Z2)
A2
````

````python
# 7. Convert probabilities to classes
y_pred = (A2 >=.5).astype(int)
y_pred
````

````python
# 8. Calculate the loss
epsilon = 1e-7

loss = -np.mean(
    y.reshape(-1, 1) * np.log(A2 + epsilon)
    +
    (1 - y.reshape(-1, 1)) * np.log(1 - A2 + epsilon)
)

print(loss)
````

## Backpropagation

````python
#For the output layer:
dZ2 = A2 - y.reshape(-1,1)
m = len(X)
dW2 = (A1.T @ dZ2) / m
db2 = np.sum(dZ2,axis=0,keepdims=True)/m
````

````python
# Then propagate backward into the hidden layer:
dA1 = dZ2 @ W2.T
#Because we used ReLU:
dZ1  = dA1 * (Z1>0)
````

````python
dW1 = (X.T @ dZ1) / m
db1 = np.sum(dZ1, axis=0, keepdims=True) / m
````

````python
# 10. Update the weights
learning_rate = 0.1

W1 -= learning_rate * dW1
b1 -= learning_rate * db1

W2 -= learning_rate * dW2
b2 -= learning_rate * db2
````

## Complete

````python
for epoch in range(10000):
    #forward
    Z1 = X @ W1 + b1
    A1 = relu(Z1)

    Z2 = A1 @ W2 + b2
    A2 = sigmoid(Z2)

    epsilon = 1e-7

    loss = -np.mean(
        y.reshape(-1, 1) * np.log(A2 + epsilon)
        +
        (1 - y.reshape(-1, 1)) * np.log(1 - A2 + epsilon)
    )

    #back
    dZ2 = A2 - y.reshape(-1,1)

    dW2 = (A1.T @ dZ2) / m
    db2 = np.sum(dZ2, axis=0, keepdims=True)/m

    dA1 = dZ2 @ W2.T
    dZ1 = dA1 * (Z1>0)

    dW1 = (X.T @ dZ1) / m
    db1 = np.sum(dZ1, axis=0, keepdims=True) / m

    #update
    W1 -= learning_rate * dW1
    b1 -= learning_rate * db1

    W2 -= learning_rate * dW2
    b2 -= learning_rate * db2
    if epoch % 1000 == 0:
        print(epoch, loss)




````

````python
import numpy as np

def softmax(Z):
    exp_Z = np.exp(Z - np.max(Z, axis=1, keepdims=True))
    return exp_Z / np.sum(exp_Z, axis=1, keepdims=True)
# ============================================================
# 1. DATA
# ============================================================

X = np.array([
    [1, 2],
    [2, 1],
    [3, 2],
    [4, 3],
    [5, 4],
    [6, 5]
])

y = np.array([
    0,
    0,
    1,
    1,
    2,
    2
])

m = len(X)


# ============================================================
# 2. FUNCTIONS
# ============================================================

def relu(Z):
    return np.maximum(0, Z)


def softmax(Z):
    exp_Z = np.exp(
        Z - np.max(Z, axis=1, keepdims=True)
    )

    return exp_Z / np.sum(
        exp_Z,
        axis=1,
        keepdims=True
    )


# ============================================================
# 3. INITIALIZE WEIGHTS AND BIASES
# ============================================================

np.random.seed(42)

# 2 input features -> 2 hidden neurons
W1 = np.random.randn(2, 2)
b1 = np.zeros((1, 2))

# 2 hidden neurons -> 3 output neurons
W2 = np.random.randn(2, 3)
b2 = np.zeros((1, 3))

learning_rate = 0.01


# ============================================================
# 4. TRAINING
# ============================================================

for epoch in range(10000):

    # -------------------------
    # FORWARD
    # -------------------------

    Z1 = X @ W1 + b1
    A1 = relu(Z1)

    Z2 = A1 @ W2 + b2
    A2 = softmax(Z2)


    # -------------------------
    # ONE-HOT ENCODE
    # -------------------------

    Y = np.zeros_like(A2)

    Y[np.arange(m), y] = 1


    # -------------------------
    # CROSS-ENTROPY LOSS
    # -------------------------

    epsilon = 1e-7

    loss = -np.mean(
        np.sum(
            Y * np.log(A2 + epsilon),
            axis=1
        )
    )


    # -------------------------
    # BACKPROP
    # -------------------------

    dZ2 = (A2 - Y) / m

    dW2 = A1.T @ dZ2
    db2 = np.sum(
        dZ2,
        axis=0,
        keepdims=True
    )

    dA1 = dZ2 @ W2.T

    dZ1 = dA1 * (Z1 > 0)

    dW1 = X.T @ dZ1
    db1 = np.sum(
        dZ1,
        axis=0,
        keepdims=True
    )


    # -------------------------
    # UPDATE
    # -------------------------

    W1 -= learning_rate * dW1
    b1 -= learning_rate * db1

    W2 -= learning_rate * dW2
    b2 -= learning_rate * db2


    if epoch % 1000 == 0:
        print(epoch, loss)


# ============================================================
# 5. PREDICTION
# ============================================================

Z1 = X @ W1 + b1
A1 = relu(Z1)

Z2 = A1 @ W2 + b2
probabilities = softmax(Z2)

predictions = np.argmax(probabilities, axis=1)

print("\nProbabilities:")
print(probabilities)

print("\nPredictions:")
print(predictions)

print("\nActual:")
print(y)
````

````python
import numpy as np

# ============================================================
# 1. DATA
# ============================================================

X = np.array([
    [1, 2],
    [2, 1],
    [3, 2]
])

y = np.array([
    [5],
    [4],
    [7]
])

m = len(X)


# ============================================================
# 2. INITIALIZE WEIGHTS AND BIASES
# ============================================================

np.random.seed(42)

# Hidden layer:
# 2 input features -> 2 hidden neurons

W1 = np.random.randn(2, 2)
b1 = np.zeros((1, 2))

# Output layer:
# 2 hidden neurons -> 1 output neuron

W2 = np.random.randn(2, 1)
b2 = np.zeros((1, 1))


learning_rate = 0.01


# ============================================================
# 3. TRAINING
# ============================================================

for epoch in range(10000):

    # --------------------------------------------------------
    # FORWARD PROPAGATION
    # --------------------------------------------------------

    Z1 = X @ W1 + b1

    A1 = np.maximum(0, Z1)       # ReLU

    Z2 = A1 @ W2 + b2

    A2 = Z2                     # Linear output for regression


    # --------------------------------------------------------
    # LOSS
    # --------------------------------------------------------

    loss = np.mean(
        (A2 - y) ** 2
    )


    # --------------------------------------------------------
    # BACKPROPAGATION
    # --------------------------------------------------------

    # MSE derivative
    dA2 = 2 * (A2 - y) / m

    # Linear activation:
    # A2 = Z2
    dZ2 = dA2

    # Output layer gradients
    dW2 = A1.T @ dZ2

    db2 = np.sum(
        dZ2,
        axis=0,
        keepdims=True
    )


    # Move backward to hidden layer
    dA1 = dZ2 @ W2.T

    # ReLU derivative
    dZ1 = dA1 * (Z1 > 0)

    # Hidden layer gradients
    dW1 = X.T @ dZ1

    db1 = np.sum(
        dZ1,
        axis=0,
        keepdims=True
    )


    # --------------------------------------------------------
    # UPDATE PARAMETERS
    # --------------------------------------------------------

    W1 -= learning_rate * dW1
    b1 -= learning_rate * db1

    W2 -= learning_rate * dW2
    b2 -= learning_rate * db2


    # --------------------------------------------------------
    # PRINT LOSS
    # --------------------------------------------------------

    if epoch % 1000 == 0:
        print(epoch, loss)


# ============================================================
# 4. PREDICTIONS
# ============================================================

Z1 = X @ W1 + b1
A1 = np.maximum(0, Z1)

Z2 = A1 @ W2 + b2
predictions = Z2

print("\nPredictions:")
print(predictions)

print("\nActual:")
print(y)
````

Yes, **almost exactly**, but one important correction:

You don't tune the model simply by comparing **training vs validation performance**. You use the **validation set to choose/tune your model and hyperparameters**, while the training set is used to **learn the weights**.

The workflow is:

```text
TRAINING SET
     ↓
learn weights
     ↓
VALIDATION SET
     ↓
evaluate → tune hyperparameters/model
     ↓
choose final model
     ↓
TEST SET
     ↓
final evaluation ONLY
```

For example:

```text
Model A → validation accuracy = 92%
Model B → validation accuracy = 94%
Model C → validation accuracy = 91%

             ↓

Choose Model B
             ↓
Evaluate Model B ONCE on test set
             ↓
Test accuracy = 93%
```

### What you should NOT do

Don't do:

```text
Test accuracy = 90%
        ↓
change hyperparameters
        ↓
Test again = 93%
        ↓
change hyperparameters
        ↓
Test again = 95%
```

Because now you're **tuning to the test set**. Your test set is no longer an unbiased estimate of how the model performs on unseen data.

### The three sets

| Dataset        | Purpose                      |
| -------------- | ---------------------------- |
| **Training**   | Learn model weights          |
| **Validation** | Choose model/hyperparameters |
| **Test**       | Final unbiased evaluation    |

So your understanding is right:

> **Train → tune using validation → freeze the choices → evaluate once on test.**

And that's why the book says **don't tweak hyperparameters based on test performance**.


Yes — **that is the standard idea in both traditional ML and deep learning.**

```text
Training set
   ↓
Learn the model's parameters/weights
   ↓
Validation set
   ↓
Tune hyperparameters + choose model
   ↓
Freeze everything
   ↓
Test set
   ↓
Estimate generalization performance
```

### What each one means

* **Training set:** the model **learns from it** (weights).
* **Validation set:** you use it to **make decisions** — hyperparameters, architecture, regularization, number of epochs, etc.
* **Test set:** you **don't make decisions from it**. You use it at the end to answer:

  > "How well do I expect this model to perform on genuinely unseen data?"

So your phrase **"test is for seeing generalization"** is exactly right.

One small correction to **"be ready for implementation"**: the test set itself doesn't make the model ready for implementation. Rather, **after you've finalized the model and evaluated it on the test set**, you have an unbiased estimate of its expected performance and can then deploy it if the result is satisfactory.

For example:

```text
Train      → learn
Validation → tune
Test       → final evaluation
Deploy     → use in the real world
```

And this applies to **ML, deep learning, CNNs, NLP models, etc.** The exact workflow can vary (e.g., cross-validation, nested CV, time-series splits), but the fundamental principle is the same.

## Building Complex Models using Functional API

About Wide and Deep neurla networks
Let's first understand what the book means by **"wide"** and **"deep."**

### 1. Regular MLP: everything goes through every layer

Imagine this MLP:

```text
Input
  ↓
Dense 1
  ↓
Dense 2
  ↓
Dense 3
  ↓
Output
```

Suppose your input is:

```text
x = [age, income, country, bought_before]
```

Every piece of information must pass through:

```text
Input → Layer 1 → Layer 2 → Layer 3 → Output
```

Each layer transforms the information.

For example:

```text
X
 ↓
XW1 + b1
 ↓
ReLU
 ↓
X2W2 + b2
 ↓
ReLU
 ↓
X3W3 + b3
 ↓
Output
```

This is great for learning **complex patterns**.

But there is a potential problem.

---

# 2. What if there is a very simple rule?

Imagine a recommendation system.

Suppose the data contains this simple relationship:

> "If the user bought this product before → recommend it again."

That's a very simple pattern.

We don't necessarily need three or ten layers of transformations to discover it.

But in a normal MLP, the information still has to go through all those layers.

```text
"bought_before"
       ↓
    Layer 1
       ↓
    Layer 2
       ↓
    Layer 3
       ↓
    Output
```

During those transformations, the network may learn useful representations, but the original simple relationship can become harder to preserve.

That's what the book means when it says:

> simple patterns may end up being distorted by this sequence of transformations

It doesn't mean the MLP **cannot** learn simple patterns. It means that forcing everything through a deep stack isn't always the most efficient architecture for preserving simple relationships.

---

# 3. Wide & Deep adds a shortcut

Wide & Deep gives the network **two paths**:

```text
                    ┌──────────────→ Output
                    │
Input ──────────────┤
                    │
                    ↓
                 Layer 1
                    ↓
                 Layer 2
                    ↓
                 Layer 3
                    ↓
                  Output
```

So there are two paths:

### Wide path

```text
Input ───────────────→ Output
```

### Deep path

```text
Input → Dense → Dense → Dense → Output
```

Hence:

**Wide + Deep**

---

# 4. Why is the wide path useful?

The wide path can learn **simple relationships** directly.

For example:

```text
bought_before = 1
        ↓
    recommend
```

The network doesn't have to transform that information through many layers.

Meanwhile, the deep path can learn complicated relationships such as:

```text
age
+
location
+
previous purchases
+
time of day
+
product category
+
user behavior
        ↓
complex representation
        ↓
recommendation
```

So you get:

```text
                 ┌── Simple patterns ──→ Output
                 │
Input ───────────┤
                 │
                 └── Complex patterns
                         ↓
                      Dense
                         ↓
                      Dense
                         ↓
                      Output
```

---

# 5. A useful analogy

Think about solving a math problem.

Suppose you have:

```text
x + 2 = 5
```

You immediately know:

```text
x = 3
```

You don't need to use a 10-step mathematical derivation.

But suppose you have:

```text
f(x) = ...
```

and the relationship is complicated.

You might need many steps.

A regular deep network is like saying:

> "Everyone must go through the 10-step process."

Wide & Deep says:

> "If there's a simple relationship, you can take the shortcut. If it's complicated, use the deep path."

---

# 6. What does "wide" actually mean?

This is important.

**Wide doesn't simply mean "a layer with lots of neurons."**

In Wide & Deep, **wide** refers to the model having a relatively direct path from features to the output.

For example:

```text
             ┌──────────── Wide ────────────┐
             │                              ↓
Input ───────┤──────────────────────────→ Output
             │
             └→ Dense → Dense → Dense ───→ Output
                         Deep
```

The wide component can also contain **manually engineered features** or feature combinations.

For example, instead of giving the network only:

```text
country = Germany
product = shoes
```

you might explicitly create:

```text
Germany × shoes
```

as a feature.

The wide component can then learn:

> "People in Germany who previously bought shoes are likely to buy these shoes."

---

# 7. How is this different from a residual connection?

This is a really important distinction because you are learning neural networks now.

You may have seen:

```text
X → Dense → Dense
 \______________→
       +
```

That's a **residual/skip connection**, as used in ResNets.

Wide & Deep is related in the general idea of **creating shortcuts**, but it isn't exactly the same architecture.

The book's point here is broader:

> Don't force every piece of information to travel through the entire deep network if some information can be useful directly.

---

# 8. The main idea to remember

You can summarize the whole paragraph like this:

```text
Regular MLP

Input
  ↓
Deep transformation
  ↓
Deep transformation
  ↓
Deep transformation
  ↓
Output

Everything goes deep.
```

Whereas:

```text
Wide & Deep

                 ┌──────────────→ Output
                 │
Input ───────────┤
                 │
                 ↓
              Deep path
                 ↓
              Deep path
                 ↓
               Output

Simple patterns + complex patterns
```

So:

**Wide path → captures simple/direct relationships**

**Deep path → captures complex/high-level relationships**

**Wide + Deep → combines both.**

And that's the reason this architecture became especially useful in **recommender systems**: recommendation often involves both simple memorized relationships ("this user bought this item") and complex generalizations ("users with these characteristics tend to like this type of item").

## 1. Normal Sequential network

With `Sequential`, your network is basically:

```text
Input
  ↓
Layer 1
  ↓
Layer 2
  ↓
Layer 3
  ↓
Output
```

Everything follows one path.

---

# 2. Functional API can create branches

For example:

```text
                 ┌→ Output 1
Input → Layers ──┤
                 └→ Output 2
```

You have **one shared network**, but it produces **two answers**.

This is where Functional API becomes very useful.

---

# Example 1 — One image, two different answers

Suppose you give the network:

```text
🐶 Image
```

You want it to answer two questions:

### Question 1: What is it?

```text
Dog
```

→ classification

### Question 2: Where is it?

```text
x = 100
y = 50
width = 200
height = 150
```

→ regression

So:

```text
                    ┌→ What? → Dog
Image → Neural Net ─┤
                    └→ Where? → [x,y,w,h]
```

You need **two outputs**.

Functional API makes this easy.

---

# Example 2 — One image, multiple tasks

Suppose the image is a person's face.

You want:

```text
Output 1 → expression
Output 2 → glasses?
```

So:

```text
                  ┌→ 😊 Smiling
Face → NN ────────┤
                  └→ 👓 Yes
```

Why use one network instead of two?

Because both tasks need similar information.

The network might learn:

```text
Face
 ↓
edges
 ↓
eyes
 ↓
mouth
 ↓
face features
```

Those features are useful for **both** tasks.

So one shared network can learn them once and use them for both outputs.

---

# Example 3 — Auxiliary output for regularization

This one is a little different.

Suppose your **main goal** is:

```text
Image → Main prediction
```

But you think the network might overfit.

You can add another output:

```text
                  ┌→ Main prediction
Image → Shared ───┤
                  └→ Auxiliary prediction
```

The auxiliary output is mainly there to **help training**.

It forces the shared layers to learn useful features instead of simply memorizing the training data.

So it can help reduce **overfitting**.

You might not even care about the auxiliary output when you're actually using the model.

---

# So what is the advantage of Functional API?

Think of the difference like this:

### Sequential

```text
A → B → C → D
```

Good for simple networks.

### Functional API

```text
        ┌→ C
A → B ──┤
        └→ D
```

or:

```text
        ┌→ C → E
A → B ──┤
        └→ D → F
```

or:

```text
A ───────────┐
             ↓
B → C → D → Combine → E
             ↑
F ───────────┘
```

You can create **branches, multiple inputs, multiple outputs, skip connections, merging, etc.**

---

## The key sentence to remember

> **Sequential API is mainly for a simple chain of layers. Functional API lets you build an arbitrary computation graph.**

That's why your earlier **Wide & Deep** model needed Functional API:

```text
                 ┌→ Wide ────────┐
Input ───────────┤               ↓
                 └→ Deep → ──→ Concatenate → Output
```

And this paragraph is showing you **another advantage**:

```text
                 ┌→ Output 1
Input → Shared ──┤
                 └→ Output 2
```

So yes: **multiple outputs and complex connections are important advantages of the Functional API.**

## Using the Subclassing API to Build Dynamic Models

````python
# build Wide and Deep NN with Subclass API
class WideAndDeepModel(tf.keras.Model):
    def __init__(self,units = 30, activation = "relu",**kwargs):
        super().__init__(**kwargs)
        self.units = units

        self.norm_layer_wide = tf.keras.layers.Normalization()
        self.norm_layer_deep = tf.keras.layers.Normalization()

        self.hidden1 = tf.keras.layers.Dense(units, activation=activation)
        self.hidden2 = tf.keras.layers.Dense(units, activation=activation)

        self.main_output = tf.keras.layers.Dense(1)
        self.aux_output = tf.keras.layers.Dense(1)

    def call(self,inputs):
        input_wide , input_deep = inputs
        norm_wide = self.norm_layer_wide(input_wide)
        norm_deep = self.norm_layer_deep(input_deep)
        hidden1 = self.hidden1(norm_deep)
        hidden2 = self.hidden2(hidden1)
        concat = tf.keras.layers.concatenate([norm_wide, hidden2])
        output = self.main_output(concat)
        aux_output = self.aux_output(hidden2)
        return output, aux_output
````

````python
model = WideAndDeepModel(units=30,activation="relu",name="my_cool_model")
````

````python
model.compile(
    loss=["mse", "mse"],
    loss_weights=[0.9, 0.1],
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    metrics=["RootMeanSquaredError"]
)
````

````python
model.norm_layer_wide.adapt(X_train_wide)
model.norm_layer_deep.adapt(X_train_deep)
````

````python
history = model.fit(
    (X_train_wide, X_train_deep),
    (y_train, y_train),
    epochs=20,
    validation_data=(
        (X_valid_wide, X_valid_deep),
        (y_valid, y_valid)
    )
)
````

````python
model.evaluate(
    (X_test_wide, X_test_deep),
    (y_test, y_test)
)
````

````python
y_pred_main, y_pred_aux = model.predict(
    (X_new_wide, X_new_deep)
)

print(y_pred_main,y_pred_aux)
````

Absolutely. This is worth understanding carefully because **Subclassing API is essentially the bridge from Keras to PyTorch-style model building**.

Let's go line by line.

---

# 1. The class definition

```python
class WideAndDeepModel(tf.keras.Model):
```

This means:

> "Create my own neural network model called `WideAndDeepModel`, based on Keras's `Model` class."

`tf.keras.Model` is the base class provided by Keras.

So you're saying:

```text
Keras Model
     ↑
     │ inherit from
     │
WideAndDeepModel
```

Because you inherit from `tf.keras.Model`, your class gets things like:

```python
model.fit()
model.predict()
model.evaluate()
model.compile()
model.save()
```

etc.

This is called **inheritance** in Python.

---

# 2. `__init__`

```python
def __init__(self, units=30, activation="relu", **kwargs):
```

This is the **constructor**.

It runs when you create the model:

```python
model = WideAndDeepModel(...)
```

Think:

> `__init__()` = "What components does my model have?"

You create the layers here.

---

## `self`

```python
def __init__(self, ...)
```

`self` means:

> "this particular model object."

For example:

```python
model = WideAndDeepModel()
```

Then:

```python
self.hidden1
```

means:

```text
model.hidden1
```

---

# 3. `units=30`

```python
units=30
```

This is just a parameter.

It means the default number of neurons in each hidden layer is 30.

So:

```python
self.hidden1 = Dense(units)
```

becomes:

```python
Dense(30)
```

unless you specify something else.

For example:

```python
model = WideAndDeepModel(units=50)
```

would create:

```text
hidden1 → 50 neurons
hidden2 → 50 neurons
```

---

# 4. `activation="relu"`

Same idea.

Default:

```python
activation="relu"
```

So:

```python
self.hidden1 = Dense(units, activation=activation)
```

means:

```python
Dense(30, activation="relu")
```

But you could do:

```python
model = WideAndDeepModel(activation="sigmoid")
```

and then the hidden layers would use sigmoid.

---

# 5. What does `**kwargs` mean?

This is the part you specifically asked about.

```python
**kwargs
```

means:

> "Accept any additional keyword arguments and collect them into a dictionary called `kwargs`."

For example:

```python
model = WideAndDeepModel(
    units=30,
    activation="relu",
    name="my_cool_model"
)
```

We didn't explicitly define:

```python
name
```

in:

```python
def __init__(self, units=30, activation="relu", **kwargs):
```

But `name="my_cool_model"` is accepted because of:

```python
**kwargs
```

Python essentially collects:

```python
kwargs = {
    "name": "my_cool_model"
}
```

---

## Why do we need this?

Because Keras's parent class (`tf.keras.Model`) accepts useful arguments such as:

```python
name
dtype
trainable
```

So we pass them to the parent.

That's what this does:

```python
super().__init__(**kwargs)
```

---

# 6. What does `super()` mean?

```python
super().__init__(**kwargs)
```

means:

> "Run the constructor of my parent class (`tf.keras.Model`)."

Remember:

```python
class WideAndDeepModel(tf.keras.Model):
```

So:

```text
WideAndDeepModel
       ↓ inherits
tf.keras.Model
```

Therefore:

```python
super().__init__()
```

calls:

```python
tf.keras.Model.__init__()
```

---

# 7. Why `**kwargs` again?

Suppose:

```python
model = WideAndDeepModel(
    name="my_cool_model"
)
```

Python receives:

```python
kwargs = {
    "name": "my_cool_model"
}
```

Then:

```python
super().__init__(**kwargs)
```

unpacks the dictionary:

```python
super().__init__(
    name="my_cool_model"
)
```

So Keras gets the name.

That's why the book says:

```python
super().__init__(**kwargs)
```

is needed to support things like naming the model.

# 8. Normalization layers

```python
self.norm_layer_wide = tf.keras.layers.Normalization()
self.norm_layer_deep = tf.keras.layers.Normalization()
```

You're creating two normalization layers.

Why two?

Because there are two inputs:

```text
Wide input → normalization
Deep input → normalization
```

And they may contain different features.

```python
self.hidden1 = tf.keras.layers.Dense(
```

Creates the first Dense layer.

With defaults:

```text
Dense(30, activation="relu")
```

Notice:

### We're only creating the layer here.

We're NOT sending data through it yet.

That's important.

# 10. Hidden layer 2

```python
self.hidden2 = tf.keras.layers.Dense(
    units,
    activation=activation
)
```

Same idea.

Again:

> Create the layer now; use it later in `call()`.

# 11. Main output

```python
self.main_output = tf.keras.layers.Dense(1)
```

One neuron.

This is the **main prediction**.

For regression:

```text
hidden representation
        ↓
Dense(1)
        ↓
prediction
```

# 12. Auxiliary output


This creates another output.

```text
                 ┌──→ main_output
                 │
hidden2 ─────────┤
                 │
                 └──→ aux_output
```

The auxiliary output can be used as an additional training signal / regularization.

# 13. Now the important part: `call()`

```python
def call(self, inputs):
```

This defines:

> **What happens to the data when it enters the model?**

Think of:

```text
__init__()
```

as:

> **Build the components**

and:

```text
call()
```

as:

> **Define the computation**

---

# 14. Two inputs

```python
input_wide, input_deep = inputs
```

The model expects:

```text
inputs = (wide_data, deep_data)
```

So Python unpacks them:

```text
inputs
 ↓
┌───────────────┐
│               │
wide            deep
```

For example:

```python
model.fit(
    (X_train_wide, X_train_deep),
    y_train
)
```

Then:

```python
input_wide = X_train_wide
input_deep = X_train_deep
```

# 15. Normalize wide input

```python
norm_wide = self.norm_layer_wide(input_wide)
```

Data flows through the wide normalization layer:

```text
input_wide
     ↓
norm_layer_wide
     ↓
norm_wide
```

---

# 16. Normalize deep input

```python
norm_deep = self.norm_layer_deep(input_deep)
```

Similarly:

```text
input_deep
     ↓
norm_layer_deep
     ↓
norm_deep
```

---

# 17. First hidden layer

```python
hidden1 = self.hidden1(norm_deep)
```

Now we're actually **using** the Dense layer we created earlier.

Data flow:

```text
norm_deep
    ↓
hidden1
```

Mathematically:

[
Z_1 = X_{deep}W_1+b_1
]

then:

[
A_1 = ReLU(Z_1)
]

---

# 18. Second hidden layer

```python
hidden2 = self.hidden2(hidden1)
```

Now:

```text
norm_deep
    ↓
hidden1
    ↓
hidden2
```

---

# 19. Concatenation

```python
concat = tf.keras.layers.concatenate(
    [norm_wide, hidden2]
)
```

This is the important Wide & Deep part.

You have:

```text
WIDE PATH:

input_wide
    ↓
normalization
    ↓
norm_wide
```

and:

```text
DEEP PATH:

input_deep
    ↓
normalization
    ↓
hidden1
    ↓
hidden2
```

Now you combine them:

```text
             norm_wide
                 │
                 │
                 ├──────→ CONCAT
                 │          ↑
                 │          │
             hidden2 ───────┘
```

The concatenation joins their features.

For example:

```text
norm_wide = [a, b, c, d, e]

hidden2 = [f, g, h, i, j, k, ...]
```

After concatenation:

```text
[a, b, c, d, e, f, g, h, i, j, k, ...]
```

Then the output layer receives **both**.

---

# 20. Main output

```python
output = self.main_output(concat)
```

Now:

```text
concat
  ↓
main_output
  ↓
output
```

This is the main prediction.

---

# 21. Auxiliary output

```python
aux_output = self.aux_output(hidden2)
```

Notice something important:

The auxiliary output uses **hidden2 directly**.

It does NOT use `concat`.

So:

```text
                    ┌──→ main_output → output
                    ↑
norm_wide → concat ─┘
               ↑
             hidden2
               ↑
             hidden1
               ↑
           norm_deep
```

And separately:

```text
hidden2
   ↓
aux_output
```

This gives the auxiliary prediction an additional training path.

---

# 22. Return both

```python
return output, aux_output
```

The model has **two outputs**:

```text
output
aux_output
```

So:

```python
y_pred = model.predict(...)
```

returns something like:

```text
(
    main_predictions,
    auxiliary_predictions
)
```

This is probably the most important thing to understand:

```text
                         INPUTS
                       /        \
                      /          \
                     ↓            ↓
                input_wide    input_deep
                     ↓            ↓
                norm_wide     norm_deep
                     │            │
                     │            ↓
                     │         hidden1
                     │            ↓
                     │         hidden2
                     │          /   \
```

```text
__init__()
    ↓
CREATE THE LAYERS

call()
    ↓
DEFINE HOW DATA FLOWS THROUGH THE LAYERS
```

That's the **core idea of the Subclassing API**.

## One final comparison

You previously built this with Functional API:

```python
input_wide = Input(...)
input_deep = Input(...)

norm_wide = norm_layer_wide(input_wide)
norm_deep = norm_layer_deep(input_deep)

hidden1 = hidden1_layer(norm_deep)
hidden2 = hidden2_layer(hidden1)

concat = concatenate([norm_wide, hidden2])

output = output_layer(concat)
```

With subclassing, you don't explicitly create `Input()` objects.

Instead:

```python
def call(self, inputs):
    input_wide, input_deep = inputs
```

The actual data comes in when you do:

```python
model.fit(
    (X_train_wide, X_train_deep),
    y_train
)
```

So the big conceptual difference is:

> **Functional API:** you explicitly construct the computation graph.

> **Subclassing API:** you create the layers in `__init__()` and write the computation yourself in `call()`.

And that's why subclassing is much more flexible for **dynamic models**.
