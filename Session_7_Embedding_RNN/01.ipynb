```python
import torch
from torch import nn, optim
```

```python
model = nn.Sequential(
    nn.Linear(1, 1),
    nn.Linear(1, 1),
    nn.Linear(1, 1)
)
```

```python
X,y = [], []
```

```python
optimizer = optim.Adam(model.parameters())
loss_fn = nn.MSELoss()
```

```python
EPOCHS = 10

for epoch in range(EPOCHS):
    model.train()
    optimizer.zero_grad()

    # Forward
    y_pred = model(X)

    # Calculate Loss
    loss = loss_fn(y, y_pred)
    loss.items()

    # Calculate Gradient
    loss.backward()


    # Update Weight
    optimizer.step()
```

```python
# Evaluate & Test
model.eval()
```
